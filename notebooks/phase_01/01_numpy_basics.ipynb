# 01 — NumPy Basics & Vector Operations
Erah AI Phase 01: Understanding vectors, matrices, broadcasting, and normalization.

```python
import numpy as np

# Vectors and Matrix Multiplication
v1 = np.array([0.5, 0.8, -0.2])
v2 = np.array([0.1, 0.9, 0.4])

print("Dot product:", np.dot(v1, v2))
print("L2 Norm:", np.linalg.norm(v1))
```
