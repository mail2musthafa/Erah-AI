# 06 — Subword Tokenization & BPE Mechanics
Erah AI Phase 01: Training Byte-Pair Encoding from scratch.

```python
from erah_ai.foundations.tokenizer import BPETokenizer

corpus = ["low lower lowest new newer newest"]
bpe = BPETokenizer(vocab_size=30)
bpe.train(corpus)

text = "lowest newer"
print("Tokens:", bpe.encode(text))
print("Decoded:", bpe.decode(bpe.encode(text)))
```
