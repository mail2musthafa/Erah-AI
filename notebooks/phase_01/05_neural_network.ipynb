Erah AI Phase 01: Forward pass, loss computation, and backprop mechanics.

```python
import numpy as np

def relu(x): return np.maximum(0, x)
def sigmoid(x): return 1 / (1 + np.exp(-x))

# 2-layer MLP forward pass
x = np.array([[1.0, 0.5]])
w1 = np.random.randn(2, 4)
w2 = np.random.randn(4, 1)

hidden = relu(np.dot(x, w1))
out = sigmoid(np.dot(hidden, w2))
print("MLP Prediction:", out)
```
